```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Basic Graph

the plot function takes two arguments x and y 
these two arguments are 1d arrays generally... 

```python
x = [1,2,3]
y = [3,4,5]
# plt.plot(x,y , color = 'r' ,linewidth = 2,   marker = '^' , markersize = 4 , linestyle = '--')
#shorthand notation : [color][marker][line]
plt.plot(x,y , 'ro:')
plt.title("A Graph in Matplotlib" , fontdict={"fontname" : "comic sans ms" , "fontsize" : 22})
plt.xlabel("The X axis" ,
          fontdict = {
              "fontname" : "times new roman",
              "fontsize" : 15
          }
          )
plt.ylabel("The Y axis",
          fontdict = {
              "fontname" : "helvetica",
              "fontsize" : 15
          }
          )

#customised x and y scales 
plt.xticks([1,2,3,4 , 3.8])
plt.yticks([3,4,5,6 , 3.2])
plt.show()
```

### Plotting two graphs at the same canvas  

```python
x = [0,1,2,3,4,5]
y = [3,4,5,6,7,10]
# plt.plot(x,y , color = 'r' ,linewidth = 2,   marker = '^' , markersize = 4 , linestyle = '--')
#shorthand notation : [color][marker][line]
plt.plot(x,y , 'ro--')
plt.title("A Graph in Matplotlib" , fontdict={"fontname" : "comic sans ms" , "fontsize" : 22})
plt.xlabel("The X axis" ,
          fontdict = {
              "fontname" : "times new roman",
              "fontsize" : 15
          }
          )
plt.ylabel("The Y axis",
          fontdict = {
              "fontname" : "helvetica",
              "fontsize" : 15
          }
          )

x2 = np.arange(0 , 5, 0.23)
plt.plot(x2 , x2**2 ,color='g')
#customised x and y scales 
# plt.xticks([1,2,3,4 , 3.8])
# plt.yticks([3,4,5,6 , 3.2])
plt.figure(figsize=(6,6), dpi = 300);
plt.savefig("graph.png" , dpi = 300)
plt.show();
```

# Bar Charts

```python
labels = ['A', 'B' , 'C']
values = [5 , 10 ,4]

bar = plt.bar(labels,values)
bar[0].set_hatch('/')
bar[1].set_hatch('o')
bar[2].set_hatch('*')
plt.show()
```
