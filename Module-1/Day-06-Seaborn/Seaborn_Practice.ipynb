### Joint Plots 

Creating a jointplot in seaborn is easy and we do it with help of `.jointplot` method 

```python
sns.jointplot(x = "tip" , y = "total_bill" , data = tips);
```

It is also easy to plot a **regression line** in this same canvas

```python
sns.jointplot(x = "tip" , y = "total_bill" , data = tips , kind = "reg")
```

```python
sns.jointplot(x = "tip" , y = "total_bill" , data = tips , kind = "kde" , shape = True , shade = True);
```

we also can denote the shading with hexagons ...

```python
sns.jointplot(x = "tip" , y = "total_bill" , data = tips , kind = "hex" ,palette = "YlGnBu" );
```

### Heatmap

Now I tried to make a heat map of titanic data set's correlation data 

```python
n = titanic.select_dtypes(include = ['number'])
sns.heatmap(n.corr() , annot = True , cmap = "YlGnBu")
```
